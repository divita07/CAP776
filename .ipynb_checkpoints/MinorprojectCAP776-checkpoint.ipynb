```python
def read_excel_file(filename):
    wb = openpyxl.load_workbook(filename, data_only=True)
    ws = wb["Daily Log"]
    
    data = []
    
    for row in ws.iter_rows(min_row=7, values_only=True):
        data.append(row)
    
    return data
```

```python
data = read_excel_file("1234567.xlsx")

print(len(data))
```

```python
from datetime import datetime

start_date = datetime(2026, 8, 13)
end_date = datetime(2026, 9, 21)

print(start_date)
print(end_date)
```

```python
def check_date(date):  #function for checking date
    if date is None:
        return False
    
    if date >= start_date and date <= end_date:
        return True
    else:
        return False
```

```python
print(check_date(datetime(2026, 8, 13)))
print(check_date(datetime(2026, 9, 21)))
print(check_date(datetime(2026, 9, 22)))
```

```python
def get_valid_records(data):
    valid_records = []
    
    for row in data:
        if check_date(row[0]):
            valid_records.append(row)
    
    return valid_records
```

```python
valid_records = get_valid_records(data)

print("Valid records:", len(valid_records))
```

```python
def calculate_average(records, column): #simple average function
    total = 0
    count = 0
    
    for row in records:
        total = total + row[column]
        count = count + 1
    
    if count == 0:
        return 0
    
    return total / count
```

```python
def calculate_tpi(records): #average coding time/day
    return calculate_average(records, 4)
```

```python
tpi = calculate_tpi(valid_records)

print("TPI =", tpi, "min/day")
```

```python
def calculate_phai(records): #average fitness time/day
    return calculate_average(records, 2)
```

```python
phai = calculate_phai(valid_records)

print("PhAI =", phai, "min/day")
```

```python
def calculate_sri(records): #average sleep time/day
    return calculate_average(records, 1)
```

```python
sri = calculate_sri(valid_records)

print("SRI =", sri, "min/day")
```

```python
def calculate_tui(records): #average total tracked time/day
    return calculate_average(records, 8)
```

```python
tui = calculate_tui(valid_records)

print("TUI =", tui, "min/day")
```

```python
def calculate_abi(records): #average free/unaccounted time/day
    return calculate_average(records, 9)
```

```python
abi = calculate_abi(valid_records)

print("ABI =", abi, "min/day")
```

```python
def calculate_aai(records): #average of study+ class time /day
    total = 0
    count = 0
    
    for row in records:
        study = row[3]
        class_time = row[5]
        
        total = total + study + class_time
        count = count + 1
    
    if count == 0:
        return 0
    
    return total / count
```

```python
aai = calculate_aai(valid_records)

print("AAI =", aai, "min/day")
```

```python
def calculate_dci(valid_days):
    expected_days = 40
    
    return (valid_days / expected_days) * 100
```

```python
dci = calculate_dci(len(valid_records))

print("DCI =", dci, "%")
```

```python
def calculate_dci(valid_days):
    expected_days = 40

    dci = (valid_days / expected_days) * 100

    return dci
```

```python
dci = calculate_dci(len(valid_records))

print("Valid Days:", len(valid_records))
print("Expected Days:", 40)
print("DCI:", dci, "%")
```

```python
print("Feeling values:")

for row in valid_records:
    print(row[10])

print("Satisfaction values:")

for row in valid_records:
    print(row[11])

print("Energy values:")

for row in valid_records:
    print(row[12])
```

```python
feeling_score = {
    "Excellent": 5,
    "Good": 4,
    "Neutral": 3,
    "Low": 2,
    "Stressed": 1
}
satisfaction_score = {
    "Very Satisfied": 5,
    "Satisfied": 4,
    "Neutral": 3,
    "Unsatisfied": 2,
    "Very Unsatisfied": 1
}
energy_score = {
    "High": 3,
    "Medium": 2,
    "Low": 1
}
```

```python
def calculate_ei(records):
    total = 0
    count = 0

    for row in records:
        feeling = feeling_score[row[10]]
        satisfaction = satisfaction_score[row[11]]
        energy = energy_score[row[12]]

        daily_score = (feeling + satisfaction + energy) / 3

        total = total + daily_score
        count = count + 1

    if count == 0:
        return 0

    ei = total / count

    return ei
```

```python
ei = calculate_ei(valid_records)

print("EI:", round(ei, 2), "/5")
```

```python
print("Data Continuity Index (DCI):", round(dci, 2), "%")
print("Experience Index (EI):", round(ei, 2), "/5")
```

```python
def calculate_pai(tpi, aai, phai, sri, tui, ei, dci):
    pai = (0.15 * tpi) + (0.20 * aai) + (0.15 * phai) + \
          (0.20 * sri) + (0.15 * tui) + (0.10 * ei) + \
          (0.05 * dci)
    
    return pai
```

```python
pai = calculate_pai(tpi, aai, phai, sri, tui, ei, dci)

print("PAI =", round(pai, 2))
```

```python
print("----- FINAL INDEX VALUES -----")

print("TPI  =", round(tpi, 2), "min/day")
print("AAI  =", round(aai, 2), "min/day")
print("PhAI =", round(phai, 2), "min/day")
print("SRI  =", round(sri, 2), "min/day")
print("ABI  =", round(abi, 2), "min/day")
print("TUI  =", round(tui, 2), "min/day")
print("EI   =", round(ei, 2), "/5")
print("DCI  =", round(dci, 2), "%")
print("PAI  =", round(pai, 2))
```

```python
for row in valid_records:
    print("Sleep:", row[1], "Energy:", row[12])
```

```python
def sleep_energy_analysis(records):
    sleep_energy = {}
    
    for row in records:
        sleep = row[1]
        energy = energy_score[row[12]]
        
        if sleep not in sleep_energy:
            sleep_energy[sleep] = []
        
        sleep_energy[sleep].append(energy)
    
    return sleep_energy
```

```python
sleep_energy = sleep_energy_analysis(valid_records)

print(sleep_energy)
```

```python
def calculate_list_average(values):
    total = 0
    
    for value in values:
        total = total + value
    
    if len(values) == 0:
        return 0
    
    return total / len(values)
```

```python
for sleep in sleep_energy:
    average_energy = calculate_list_average(sleep_energy[sleep])
    
    print("Sleep:", sleep, "minutes")
    print("Average Energy:", round(average_energy, 2), "/3")
    print()
```
